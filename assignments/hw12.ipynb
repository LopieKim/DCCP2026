```python
import numpy as np
import pandas as pd

works = pd.DataFrame({
    '작가': ['김유정', '김유정', '김유정', '현진건', '현진건', '이상', '이상', '나도향', '나도향', '채만식', '이효석'],
    '작품': ['봄봄', '동백꽃', '만무방', '운수 좋은 날', 'B사감과 러브레터', '날개', '권태', '벙어리 삼룡이', '물레방아', '레디메이드 인생', '메밀꽃 필 무렵'],
    '연도': [1935, 1936, 1934, 1924, 1925, 1936, 1937, 1925, 1925, 1934, np.nan],
    '글자수': [12500, 9800, 13600, 11200, 7400, 18300, 14100, 16800, 14500, 22000, 11800],
})

authors = pd.DataFrame({
    '작가': ['김유정', '현진건', '이상', '나도향', '채만식', '이효석', '염상섭'],
    '생년': [1908, 1900, 1910, 1902, 1902, 1907, 1897],
    '몰년': [1937, 1943, 1937, 1926, 1950, 1942, 1963],
})
```

Q1 (a)

```python
print("행과 열의 개수")
print(works.shape)

print("\n각 열의 자료형")
print(works.dtypes)

print("숫자 열의 요약 통계")
print(works.describe())
```

Q1(a) 설명: 연도 열에 NaN이 하나 포함되어 있기 때문에 열 전체의 자료형이 float64로 변환됩니다.

Q1(b)

```python
print(works.isna().sum())
```

Q1(b) 설명: isna(sum)으로 각 열의 결측치의 개수를 확인합니다.

Q1(c)

```python
works2 = works.fillna({'연도': 1936})
works2['연도'] = works2['연도'].astype(int)

print(works2.dtypes)
print(works2.tail(3))
```

Q1(c) 설명: fillna()로 결측값을 1936으로 채웠습니다. 이후 astype(int)로 형변환을 했습니다.

Q2(a)

```python
after_1930 = works2[works2['연도'] > 1930]
print("1930년 이후 작품")
print(after_1930)
kim_or_lee = works2[works2['작가'].isin(['김유정', '이상'])]
print("\n김유정 또는 이상 작품")
print(kim_or_lee)
```

Q2(a) 설명: isin을 사용해서 원하는 조건의 값들을 추출했습니다.

Q2(b)

```python
def categorize(n: int) -> str:
    if n < 10000:
        return '짧음'
    elif n <= 15000:
        return '보통'
    else:
        return '긴'

works2['분량'] = works2['글자수'].apply(categorize)
print(works2[['작품', '글자수', '분량']])
```

Q2(b) 설명: categorize라는 함수를 만들어서 분량이라는 열을 추가했습니다.

Q3(a)

```python
author_stats = works2.groupby('작가')['글자수'].agg(['mean', 'count', 'sum'])
print(author_stats)
```

Q3(a) 설명: 작가별 평균, 작품수, 글자수를 구했습니다.

Q3(b)

```python
length_counts = works2['분량'].value_counts()
print(length_counts)
```

Q3(b) 설명: value_counts로 works2의 분량별 작품수를 구했습니다.

Q3(c)

```python
author_length_counts = works2.groupby(['작가', '분량']).size()
print(author_length_counts)
```

Q3(c) 설명: 가장 많은 작품 수를 가진 조합은 ('김유정', '보통')으로 총 2건입니다. 이는 김유정 작가가 주로 10,000자에서 15,000자 사이 글자수의 글을 썼다는 것입니다.

Q4(a)

```python
full = pd.merge(works2, authors, on='작가', how='left')
print(full)
print(f"\n합쳐진 표의 shape: {full.shape}")
```

Q4(a) 설명: how = left 조건은 왼쪽 표(works2)에 존재하는 key(작가)를 기준으로 데이터를 병합하므로 full에 포함되지 않습니다. 

Q4(b)

```python
full['집필 당시 나이'] = full['연도'] - full['생년']
print(full[['작가', '작품', '연도', '생년', '집필 당시 나이']])
```

Q4(b) 설명: 집필 당시 나이 열을 추가했습니다.

Q4(c)

```python
avg_age_by_author = full.groupby('작가')['집필 당시 나이'].mean().sort_values()
print(avg_age_by_author)
```

Q4(c) 설명: 가장 어린 나이에 작품을 발표한 작가는 '나도향'이며, 평균 23.0세에 작품을 썼습니다. 반면 가장 늦은 나이(평균 32.0세)에 작품을 발표한 작가는 '채만식'입니다.
